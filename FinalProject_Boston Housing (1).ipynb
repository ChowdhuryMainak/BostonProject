#### Read the dataset in the csv file from the URL


```python
from js import fetch
import io

URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/boston_housing.csv'
resp = await fetch(URL)
boston_url = io.BytesIO((await resp.arrayBuffer()).to_py())
```

```python
boston_df=pd.read_csv(boston_url)
```

```python
if 'Unnamed: 0' in boston_df.columns:
    boston_df = boston_df.drop(columns=['Unnamed: 0'])
    
```

```python
display(boston_df.head())
display(boston_df.dtypes)
display(boston_df.isnull().sum())
print("Number of duplicate rows:", boston_df.duplicated().sum())
display(boston_df.corr(numeric_only=True))
display(boston_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=boston_df, x='INDUS', y='NOX')
plt.title('Nitric Oxide Concentration vs Non-Retail Business Acres')
plt.xlabel('INDUS: Non-retail business acres per town')
plt.ylabel('NOX: Nitric oxide concentration')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.histplot(data=boston_df, x='PTRATIO', bins=15, kde=True)
plt.title('Distribution of Pupil-Teacher Ratio')
plt.xlabel('PTRATIO')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(y=boston_df['MEDV'])
plt.title('Boxplot of Median Home Value (MEDV)')
plt.ylabel('MEDV ($1,000s)')
plt.tight_layout()
plt.show()
```

```python
boston_df['AGE_group'] = pd.cut(
    boston_df['AGE'],
    bins=[-float('inf'), 35, 70, float('inf')],
    labels=['35 and younger', 'Between 35 and 70', '70 and older'],
    include_lowest=True
)
plt.figure(figsize=(8, 5))
sns.boxplot(data=boston_df, x='AGE_group', y='MEDV')
plt.title('Median Home Value by AGE Group')
plt.xlabel('Proportion of owner-occupied units built before 1940 (AGE group)')
plt.ylabel('MEDV ($1,000s)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=boston_df, x='CHAS')
plt.title('Homes by Charles River Boundary (CHAS)')
plt.xlabel('CHAS (1 = bounds Charles River, 0 = does not)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
